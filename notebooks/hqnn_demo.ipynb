# Hybrid Quantum–Classical Neural Network Demo
This notebook shows how a small classical neural network can feed parameters into a PennyLane quantum circuit, followed by a classical classifier. All plots are rendered as SVG for easy viewing on GitHub.

```python
%config InlineBackend.figure_formats = ['svg']
import matplotlib.pyplot as plt
import torch
from app.dataset import load_moons
from app.hybrid_model import HybridModel
from app.trainer import Trainer
from app.plots import plot_decision_boundary, plot_training_loss, plot_latent_space
```

## Load data and build the model

```python
train_loader, test_loader, X_train, y_train = load_moons()
model = HybridModel()
trainer = Trainer(model, lr=0.01)
```

## Train the hybrid model

```python
history = trainer.train(train_loader, test_loader, epochs=5)
history
```

## Visualize results

```python
# Encoder outputs for latent space projection
with torch.no_grad():
    encoder_outputs = model.encoder(torch.tensor(X_train, dtype=torch.float32)).numpy()

plot_decision_boundary(model, X_train, y_train, '../examples/decision_boundary_hybrid.svg')
plot_training_loss(history, '../examples/training_loss.svg')
plot_latent_space(encoder_outputs, '../examples/latent_space_visualization.svg')

from IPython.display import SVG, display
display(SVG(filename='../examples/decision_boundary_hybrid.svg'))
display(SVG(filename='../examples/training_loss.svg'))
display(SVG(filename='../examples/latent_space_visualization.svg'))
```
